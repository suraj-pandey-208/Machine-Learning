<!-- PIE CHART STARTS FROM HERE -->

```python


import matplotlib.pyplot as plt

values = [50, 30, 15, 5]
data = ["Rent", "Food", "Travel", "Other"]

plt.pie(
    values,
    
    labels=data,
    colors=["blue", "black", ""
    "white", "green"],

    autopct="%1.2f%%",
    startangle=0,
    explode=None,
    shadow=False,
    counterclock=True,
    wedgeprops=None,
    textprops=None,
    radius=1,
    center=(0, 0),
    frame=False,
    labeldistance=1.1,
    pctdistance=0.6
)

plt.show()
```

```python
import pandas as pd;
sale_data_df=pd.read_csv("../sale_data.csv")
sale_data_df.info()
```

```python

# PLOT THE TOTAL SALES BY PRODUCT

total_sale_by_product=sale_data_df.groupby('Product Category')['Total'].sum()
print(total_sale_by_product)

total_sale_by_product.plot(kind='bar', color='teal')
```

```python
# PLOT SALES TREND OVER TIME
sales_trend=sale_data_df.groupby('Date')['Total'].sum().reset_index()
print(sales_trend)
# sales_trend.plot(kind='bar',color="red")
```

```python
# plt.bar( sales_trend.index,sales_trend.values)
plt.bar( sales_trend['Date'],sales_trend['Total'])



plt.show()
```
